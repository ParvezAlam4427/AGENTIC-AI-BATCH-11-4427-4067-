```python
!pip install --upgrade gradio google-genai --quiet

from google import genai
import gradio as gr
import os

# -------------------------------
# Configuration
# -------------------------------
GEMINI_API_KEY = os.environ.get("GEMINI_API_KEY", "")
if GEMINI_API_KEY:
    os.environ["GEMINI_API_KEY"] = GEMINI_API_KEY
    gemini_client = genai.Client()
else:
    gemini_client = None

# -------------------------------
# Authentication
# -------------------------------
users_db = {}
logged_in_user = [None]

def signup(username, password):
    if username in users_db:
        return "❌ Username already exists!"
    users_db[username] = password
    return "✅ Signup successful! Please login."

def login(username, password):
    if username in users_db and users_db[username] == password:
        logged_in_user[0] = username
        return True, f"✅ Welcome, {username}!"
    return False, "❌ Invalid username or password."

# -------------------------------
# Topic Suggestion
# -------------------------------
def suggest_trending(profile_text):
    keywords = (profile_text or "").lower()
    suggestions = []
    if "health" in keywords or "medical" in keywords:
        suggestions.append("AI in Healthcare")
    if "nlp" in keywords or "language" in keywords:
        suggestions.append("Natural Language Processing (NLP)")
    if "robot" in keywords or "automation" in keywords:
        suggestions.append("Robotics & Automation")
    if "quantum" in keywords:
        suggestions.append("Quantum Computing")
    if "energy" in keywords or "green" in keywords:
        suggestions.append("Green Energy AI")
    if "vision" in keywords or "image" in keywords:
        suggestions.append("Computer Vision")
    if "data" in keywords:
        suggestions.append("Data Science and Predictive Analytics")
    return suggestions if suggestions else ["General AI Research"]

# -------------------------------
# Main Assistant Function
# -------------------------------
def agentic_ai_assistant(faculty_profile, topic_choice, custom_topic, dark_mode=False):
    topic = (custom_topic.strip() if custom_topic and custom_topic.strip() else topic_choice) or "General AI Research"
    query = f"Faculty profile: {faculty_profile}\nFocus topic: {topic}"

    if gemini_client:
        try:
            summary_response = gemini_client.models.generate_content(
                model="gemini-2.5-flash",
                contents=f"Generate a detailed theoretical research overview on: {query}"
            )
            research_summary = summary_response.text
        except Exception as e:
            research_summary = f"⚠ Gemini API Error: {e}"

        try:
            insights_response = gemini_client.models.generate_content(
                model="gemini-2.5-flash",
                contents=f"Provide grants, collaborations, and latest trends for: {query}"
            )
            gemini_insights = insights_response.text
        except Exception as e:
            gemini_insights = f"⚠ Gemini API Error: {e}"
    else:
        topic_lower = topic.lower()

        # -------------------------------
        # EXTENDED DETAILED THEORY SECTION
        # -------------------------------
        if "machine learning" in topic_lower:
            research_summary = """
            <h3>📘 Machine Learning — Extensive Overview</h3>
            <p><b>Definition:</b> Machine Learning (ML) is a subfield of Artificial Intelligence (AI) focused on developing systems that can automatically learn and improve from experience without explicit programming.</p>

            <h4>🔹 History & Evolution</h4>
            <p>Machine Learning originated from pattern recognition and computational learning theory. Early work by Alan Turing (1950) on learning machines laid the foundation for modern ML. The field matured in the 1990s with the rise of neural networks and data-driven algorithms.</p>

            <h4>⚙️ Types of Learning</h4>
            <ul>
                <li><b>Supervised Learning:</b> Uses labeled datasets (e.g., regression, classification).</li>
                <li><b>Unsupervised Learning:</b> Finds patterns in unlabeled data (e.g., clustering, association).</li>
                <li><b>Semi-supervised Learning:</b> Combines small labeled data with large unlabeled data.</li>
                <li><b>Reinforcement Learning:</b> Agent learns via trial and reward-based feedback.</li>
            </ul>

            <h4>🧮 Common Algorithms</h4>
            <ul>
                <li>Linear & Logistic Regression</li>
                <li>Decision Trees & Random Forests</li>
                <li>Support Vector Machines (SVM)</li>
                <li>Gradient Boosting (XGBoost, LightGBM)</li>
                <li>K-Means & Hierarchical Clustering</li>
            </ul>

            <h4>🌍 Real-world Applications</h4>
            <ul>
                <li>Healthcare: Disease prediction & medical imaging</li>
                <li>Finance: Credit scoring, fraud detection</li>
                <li>Education: Student performance analytics</li>
                <li>Transportation: Traffic forecasting, autonomous driving</li>
            </ul>

            <h4>🚀 Current Research Directions</h4>
            <ul>
                <li>Explainable AI (XAI)</li>
                <li>Green and Sustainable Machine Learning</li>
                <li>Federated and Privacy-Preserving Learning</li>
            </ul>
            """

        elif "random forest" in topic_lower:
            research_summary = """
            <h3>🌳 Random Forest — Detailed Research Theory</h3>
            <p><b>Definition:</b> Random Forest is an ensemble method that combines multiple decision trees to produce a more accurate and stable prediction.</p>

            <h4>🔹 Algorithm Workflow</h4>
            <ul>
                <li>Creates multiple bootstrapped datasets.</li>
                <li>Each tree randomly selects a subset of features.</li>
                <li>Predictions are aggregated via majority voting or averaging.</li>
            </ul>

            <h4>📈 Mathematical Representation</h4>
            <p>Given trees T₁, T₂, ..., Tₙ trained on bootstrapped data, the Random Forest prediction is:</p>
            <pre style="background:#f3f4f6; padding:8px;">ŷ = majority_vote(T₁(x), T₂(x), ..., Tₙ(x))</pre>

            <h4>⚙️ Advantages</h4>
            <ul>
                <li>High accuracy and robustness to overfitting.</li>
                <li>Handles large datasets and missing values.</li>
                <li>Provides interpretable feature importance scores.</li>
            </ul>

            <h4>🌍 Applications</h4>
            <ul>
                <li>Medical diagnosis (predicting cancer outcomes)</li>
                <li>Finance (credit risk analysis)</li>
                <li>Marketing (customer segmentation)</li>
                <li>Cybersecurity (intrusion detection)</li>
            </ul>

            <h4>🧠 Ongoing Research</h4>
            <ul>
                <li>Hybrid Random Forest with Neural Networks</li>
                <li>Energy-efficient Random Forest implementations</li>
                <li>Interpretable ensemble models</li>
            </ul>
            """

        elif "deep learning" in topic_lower:
            research_summary = """
            <h3>🧬 Deep Learning — Theoretical Expansion</h3>
            <p><b>Definition:</b> Deep Learning (DL) uses multi-layer neural networks to learn high-level abstractions from data.</p>
            <h4>🔹 Core Architectures</h4>
            <ul>
                <li><b>CNNs:</b> Used for image analysis and computer vision.</li>
                <li><b>RNNs / LSTMs:</b> Used for time-series and sequence modeling.</li>
                <li><b>Transformers:</b> Foundation of modern large language models (LLMs).</li>
            </ul>
            <h4>⚙️ Applications</h4>
            <ul>
                <li>Speech Emotion Recognition</li>
                <li>Autonomous Driving</li>
                <li>Medical Image Segmentation</li>
                <li>Natural Language Understanding</li>
            </ul>
            """

        elif "nlp" in topic_lower or "language" in topic_lower:
            research_summary = """
            <h3>💬 Natural Language Processing (NLP) — In-depth Overview</h3>
            <p>NLP bridges the gap between human communication and computer understanding. It integrates linguistics, statistics, and deep learning.</p>
            <h4>🔹 Major Areas</h4>
            <ul>
                <li>Text Classification, Sentiment Analysis</li>
                <li>Machine Translation</li>
                <li>Speech Recognition & Chatbots</li>
                <li>Text Summarization</li>
            </ul>
            <h4>🧠 Modern Models</h4>
            <ul>
                <li>BERT (Bidirectional Encoder Representations)</li>
                <li>GPT and Transformer architectures</li>
                <li>Gemini and Multimodal AI</li>
            </ul>
            """

        else:
            research_summary = f"""
            <h3>📚 Research Overview on {topic}</h3>
            <p>This topic explores interdisciplinary advancements, real-world applications, and the future of AI-driven innovation.</p>
            """

        gemini_insights = """
        <h4>🤝 Collaboration & Grant Opportunities</h4>
        <ul>
            <li>Collaborate with research institutes (IITs, MIT, Stanford).</li>
            <li>Apply for AICTE, DST, and Horizon Europe funding programs.</li>
            <li>Partner with industries for applied AI solutions.</li>
        </ul>
        """

    # -------------------------------
    # VISUALLY CLEAR OUTPUT
    # -------------------------------
    html_output = f"""
    <div style="background:#ffffff; color:#0f172a; padding:22px; border-radius:14px; border:1px solid #cbd5e1;
                line-height:1.7; font-size:16px;">
        <h2 style="color:#1e3a8a;">📊 Research Topic: {topic}</h2>
        {research_summary}
        {gemini_insights}
        <h4>💡 Future Scope</h4>
        <ul>
            <li>Integration of Explainable AI (XAI) frameworks.</li>
            <li>Use of cloud-based distributed learning systems.</li>
            <li>Ethical AI and fairness-driven models.</li>
        </ul>
    </div>
    """
    return html_output

# -------------------------------
# Gradio Interface
# -------------------------------
with gr.Blocks(css="""
body {
    font-family: 'Poppins', sans-serif;
    background-color: #e2e8f0;
}
.card {
    border-radius: 14px;
    padding: 18px;
    box-shadow: 0 6px 18px rgba(0,0,0,0.1);
    background: #ffffff;
}
""") as app:

    gr.Markdown("## ⚡ **AI Research Assistant — Faculty Research Hub**")

    auth_card = gr.Group(visible=True)
    with auth_card:
        with gr.Column(elem_classes="card"):
            gr.Markdown("### 🔐 Login / Signup")
            mode = gr.Radio(["Login", "Signup"], value="Login", label="Select Mode")
            user = gr.Textbox(label="Username")
            passwd = gr.Textbox(label="Password", type="password")
            auth_btn = gr.Button("Continue 🚀")
            auth_status = gr.Label()

    research_section = gr.Group(visible=False)
    with research_section:
        with gr.Row():
            with gr.Column(scale=1, elem_classes="card"):
                gr.Markdown("### 👤 Faculty Profile")
                profile_input = gr.Textbox(lines=6, placeholder="Enter your research profile...")
                suggest_btn = gr.Button("✨ Suggest Trending Topics")
                trending_dropdown = gr.Dropdown(choices=[], label="Trending Topics")
                custom_topic = gr.Textbox(label="Or enter a custom topic (optional)")
                submit_btn = gr.Button("🚀 Generate Detailed Research")
                logout_btn = gr.Button("🔓 Logout")
            with gr.Column(scale=2, elem_classes="card"):
                gr.Markdown("### 📚 Research Output")
                output_html = gr.HTML()

    def auth_action(mode, username, password):
        if mode == "Login":
            success, message = login(username, password)
            return message, gr.update(visible=not success), gr.update(visible=success)
        else:
            msg = signup(username, password)
            return msg, gr.update(), gr.update()

    def suggest_topics(profile_text):
        topics = suggest_trending(profile_text)
        return gr.update(choices=topics, value=topics[0])

    def logout_action():
        logged_in_user[0] = None
        return gr.update(visible=True), gr.update(visible=False)

    auth_btn.click(fn=auth_action, inputs=[mode, user, passwd], outputs=[auth_status, auth_card, research_section])
    suggest_btn.click(fn=suggest_topics, inputs=[profile_input], outputs=[trending_dropdown])
    submit_btn.click(fn=agentic_ai_assistant, inputs=[profile_input, trending_dropdown, custom_topic], outputs=[output_html])
    logout_btn.click(fn=logout_action, outputs=[auth_card, research_section])

app.launch()
```
